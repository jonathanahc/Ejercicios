# **Clase 9 — Sobrecarga de funciones y operadores**


## **1. ¿Qué es la Sobrecarga?**

La **sobrecarga** (*overloading*) consiste en utilizar un mismo nombre para representar funciones u operaciones diferentes, siempre que el lenguaje pueda distinguirlas.

Estudiaremos dos formas principales:

- **Sobrecarga de funciones:** varias funciones comparten nombre, pero tienen diferentes listas de parámetros.
- **Sobrecarga de operadores:** se define el comportamiento de determinados operadores cuando se aplican a objetos de clases o estructuras creadas por el programador.

## ¿Para qué sirve?

- Permite usar nombres consistentes para operaciones relacionadas.
- Hace que las interfaces de las clases sean más naturales.
- Reduce la necesidad de crear nombres como `sumarDos`, `sumarTres` y `sumarDecimales`.
- Facilita la lectura del código si el significado de cada operación está bien definido.

**Importante:** sobrecargar no significa repetir exactamente una función. Cada versión debe poder distinguirse de las demás.

# **2. Sobrecarga de funciones**

Una función puede sobrecargarse cuando sus versiones tienen el mismo nombre, pero una lista de parámetros diferente.

La lista puede diferenciarse por:
1. **Cantidad:** dos parámetros frente a tres.
2. **Tipos:** `int` frente a `double`.
3. **Orden de tipos:** por ejemplo, `(int, double)` frente a `(double, int)`.

El tipo de retorno no basta para distinguir funciones. Estas declaraciones son inválidas como sobrecarga:
```cpp
int obtener(int x);
double obtener(int x); // Error: mismos parámetros
```

El compilador analiza los argumentos de la llamada y selecciona la versión más adecuada. Si ninguna versión es viable o hay ambigüedad, se produce un error de compilación.

### **2.1 Sobrecarga por cantidad de parámetros**

En este caso, ambas funciones se llaman `sumar`, pero una recibe dos enteros y la otra recibe tres. La cantidad de argumentos permite distinguirlas.

**Qué observar:** `sumar(4, 6)` llama a la versión de dos parámetros; `sumar(4, 6, 8)` llama a la versión de tres.

````python
%%writefile funciones_cantidad.cpp
#include <iostream>
using namespace std;

int sumar(int a, int b) {
    return a + b;
}

int sumar(int a, int b, int c) {
    return a + b + c;
}

int main() {
    cout << "Suma de dos numeros: " << sumar(4, 6) << '\n';
    cout << "Suma de tres numeros: " << sumar(4, 6, 8) << '\n';
    return 0;
}
````

````python
!g++ funciones_cantidad.cpp -o funciones_cantidad && ./funciones_cantidad
````

### **2.2 Sobrecarga por tipo de dato**

Aquí las funciones reciben la misma cantidad de parámetros, pero sus tipos son distintos. Una versión trabaja con valores `int` y la otra con valores `double`.

- `int` se usa para números enteros.
- `double` se usa para números de punto flotante.
- Los literales `2.5` y `4.8` son de tipo `double`, por lo que la llamada corresponde a esa versión.

````python
%%writefile funciones_tipo.cpp
#include <iostream>
using namespace std;

int sumar(int a, int b) {
    return a + b;
}

double sumar(double a, double b) {
    return a + b;
}

int main() {
    cout << "Enteros: " << sumar(5, 3) << '\n';
    cout << "Decimales: " << sumar(2.5, 4.8) << '\n';
    return 0;
}
````

````python
!g++ funciones_tipo.cpp -o funciones_tipo && ./funciones_tipo
````

### **2.3 Sobrecarga con distintos tipos de información**

La sobrecarga no se limita a operaciones aritméticas. También puede usarse para crear una función que muestre distintos tipos de datos.

En este ejemplo, el compilador distingue las versiones de `mostrar` según si recibe un entero, un decimal o una cadena de texto.

````python
%%writefile funciones_mostrar.cpp
#include <iostream>
#include <string>
using namespace std;

void mostrar(int dato) {
    cout << "Entero: " << dato << '\n';
}

void mostrar(double dato) {
    cout << "Decimal: " << dato << '\n';
}

void mostrar(const string& dato) {
    cout << "Texto: " << dato << '\n';
}

int main() {
    mostrar(25);
    mostrar(18.75);
    mostrar(string("Programacion Orientada a Objetos"));
    return 0;
}
````

## **2.4 ¿Cómo resuelve el compilador la sobrecarga?**

El compilador compara los argumentos de la llamada con los parámetros de las funciones disponibles. En términos generales, prefiere la coincidencia más directa y aplica las reglas de resolución de sobrecarga de C++.

| Declaraciones disponibles | Llamada | Versión esperada |
|---|---|---|
| `sumar(int, int)` y `sumar(double, double)` | `sumar(2, 3)` | `sumar(int, int)` |
| Las mismas | `sumar(2.5, 3.5)` | `sumar(double, double)` |
| `sumar(int, int)` y `sumar(int, int, int)` | `sumar(2, 3, 4)` | `sumar(int, int, int)` |

**Atención:** las conversiones implícitas pueden hacer que algunos casos sean más complejos. Si dos opciones resultan igualmente viables, la llamada puede ser ambigua.

### **2.5 Aplicación: calcular costos**

Una empresa puede calcular el costo de uno, dos o tres servicios. La sobrecarga permite conservar el nombre `calcularCosto` y cambiar la cantidad de servicios que se envían como argumentos.

Este ejemplo combina el concepto de sobrecarga por cantidad de parámetros con una situación cotidiana.

````python
%%writefile calcular_costos.cpp
#include <iostream>
using namespace std;

double calcularCosto(double a) {
    return a;
}

double calcularCosto(double a, double b) {
    return a + b;
}

double calcularCosto(double a, double b, double c) {
    return a + b + c;
}

int main() {
    cout << "Un servicio: " << calcularCosto(150.0) << '\n';
    cout << "Dos servicios: " << calcularCosto(150.0, 200.0) << '\n';
    cout << "Tres servicios: " << calcularCosto(150.0, 200.0, 100.0) << '\n';
    return 0;
}
````

# **3. Sobrecarga de operadores**

Los operadores como `+`, `-` y `==` tienen significados definidos para tipos incorporados, como `int` y `double`. Para una clase propia, C++ no sabe automáticamente qué significa sumar o comparar dos objetos.

La **sobrecarga de operadores** permite programar ese comportamiento.

Por ejemplo, si `Vector2D` representa un vector con coordenadas `x` e `y`, podemos decidir que:
- `v1 + v2` sume las coordenadas correspondientes.
- `v1 - v2` reste las coordenadas correspondientes.
- `v1 == v2` compruebe si ambas coordenadas coinciden.

La implementación debe respetar el significado que tenga sentido para la clase. No conviene sobrecargar un operador de forma que confunda a quien lea el programa.

## Sintaxis habitual de una función miembro

```cpp
TipoRetorno operatorSIMBOLO(parametros) const {
    // comportamiento definido por el programador
}
```

Por ejemplo, `operator+` define el comportamiento del operador `+`. El modificador `const` al final indica que la función no pretende modificar el objeto actual.

### **3.1 Sobrecarga de `+`: sumar vectores**

La clase `Vector2D` representa un vector con dos coordenadas. La función `operator+` recibe el segundo vector y devuelve un nuevo vector con la suma de ambas coordenadas.

En `Vector2D resultado = v1 + v2`, C++ utiliza la implementación de `operator+` definida en la clase.

````python
%%writefile operador_suma.cpp
#include <iostream>
using namespace std;

class Vector2D {
public:
    double x, y;

    Vector2D(double x, double y) : x(x), y(y) {}

    Vector2D operator+(const Vector2D& otro) const {
        return Vector2D(x + otro.x, y + otro.y);
    }

    void mostrar() const {
        cout << "(" << x << ", " << y << ")" << '\n';
    }
};

int main() {
    Vector2D v1(2, 3);
    Vector2D v2(4, 5);
    Vector2D resultado = v1 + v2;

    cout << "Vector 1: "; v1.mostrar();
    cout << "Vector 2: "; v2.mostrar();
    cout << "Suma: "; resultado.mostrar();
    return 0;
}
````

### **3.2 Sobrecarga de `-`: restar vectores**

En esta clase, la resta de dos vectores se define restando sus coordenadas correspondientes. La operación devuelve un objeto nuevo y no modifica los vectores originales.

La expresión `a - b` es binaria porque utiliza dos operandos.

````python
%%writefile operador_resta.cpp
#include <iostream>
using namespace std;

class Vector2D {
public:
    int x, y;

    Vector2D(int x, int y) : x(x), y(y) {}

    Vector2D operator-(const Vector2D& otro) const {
        return Vector2D(x - otro.x, y - otro.y);
    }

    void mostrar() const {
        cout << "(" << x << ", " << y << ")" << '\n';
    }
};

int main() {
    Vector2D a(10, 8);
    Vector2D b(3, 2);
    Vector2D diferencia = a - b;

    cout << "Diferencia: ";
    diferencia.mostrar();
    return 0;
}
````

# **4. Sobrecarga de operadores relacionales**

Los operadores relacionales permiten comparar objetos. La clase debe definir qué significa comparar sus instancias.

- `==`: comprueba igualdad.
- `<`: comprueba si un objeto es menor que otro según un criterio.
- `>`: comprueba si un objeto es mayor que otro según un criterio.

El criterio depende del problema. En una clase `Punto`, `==` puede comparar coordenadas. En una clase `Producto`, `<` puede comparar precios. La comparación debe ser coherente y documentarse cuando su significado no sea obvio.

### **4.1 Sobrecarga de `==`: comparar puntos**

En este ejemplo, dos puntos se consideran iguales solamente si sus coordenadas `x` e `y` son iguales.

La función devuelve `bool`, es decir, un valor lógico: `true` o `false`. `const Punto&` evita copiar el objeto comparado y declara que no se modificará a través de ese parámetro.

````python
%%writefile operador_igual.cpp
#include <iostream>
using namespace std;

class Punto {
public:
    int x, y;

    Punto(int x, int y) : x(x), y(y) {}

    bool operator==(const Punto& otro) const {
        return x == otro.x && y == otro.y;
    }
};

int main() {
    Punto p1(4, 7);
    Punto p2(4, 7);
    Punto p3(2, 9);

    cout << boolalpha;
    cout << "p1 == p2: " << (p1 == p2) << '\n';
    cout << "p1 == p3: " << (p1 == p3) << '\n';
    return 0;
}
````

### **4.2 Sobrecarga de `<` y `>`: comparar productos**

La clase `Producto` tiene un nombre y un precio. Para este problema, dos productos se comparan por su precio, no por su nombre.

La sobrecarga de `<` y `>` hace posible escribir expresiones legibles como `producto1 < producto2`.

````python
%%writefile operadores_menor_mayor.cpp
#include <iostream>
#include <string>
using namespace std;

class Producto {
public:
    string nombre;
    double precio;

    Producto(string nombre, double precio)
        : nombre(nombre), precio(precio) {}

    bool operator<(const Producto& otro) const {
        return precio < otro.precio;
    }

    bool operator>(const Producto& otro) const {
        return precio > otro.precio;
    }
};

int main() {
    Producto p1("Teclado", 450.0);
    Producto p2("Monitor", 3200.0);

    cout << boolalpha;
    cout << p1.nombre << " cuesta menos que " << p2.nombre
         << ": " << (p1 < p2) << '\n';
    cout << p2.nombre << " cuesta mas que " << p1.nombre
         << ": " << (p2 > p1) << '\n';
    return 0;
}
````

# **5. Operadores unarios**

Un operador **unario** actúa sobre un solo operando. Algunos ejemplos son `-x`, `++x` y `--x`.

En una clase, podemos definir qué significa aplicar un operador unario a un objeto.

Es importante distinguir:
- `a - b`: `-` binario, porque hay dos operandos.
- `-a`: `-` unario, porque hay un solo operando.
- `++c`: incremento de prefijo.
- `c++`: incremento de posfijo.

Aunque algunos símbolos se repiten, la firma de la función de sobrecarga distingue el comportamiento correspondiente.

### **5.1 Sobrecarga del operador unario `-`**

En un vector, definiremos el operador unario `-` como el vector con las coordenadas de signo contrario. La función no recibe parámetros explícitos porque el operando es el objeto actual (`*this`).

Esta operación devuelve un nuevo vector y no modifica el original.

````python
%%writefile operador_unario_menos.cpp
#include <iostream>
using namespace std;

class Vector2D {
public:
    int x, y;

    Vector2D(int x, int y) : x(x), y(y) {}

    Vector2D operator-() const {
        return Vector2D(-x, -y);
    }

    void mostrar() const {
        cout << "(" << x << ", " << y << ")" << '\n';
    }
};

int main() {
    Vector2D v(4, -6);
    Vector2D opuesto = -v;

    cout << "Original: "; v.mostrar();
    cout << "Opuesto: "; opuesto.mostrar();
    return 0;
}
````

### **5.2 Sobrecarga de `++` y `--`**


`++` incrementa y `--` decrementa el valor de un objeto, según el comportamiento que defina la clase.

**Prefijo:** `++c` modifica primero el objeto y la expresión representa el valor actualizado.  
**Posfijo:** `c++` conserva una copia del valor anterior como resultado de la expresión y luego incrementa el objeto.

Para distinguir ambas sobrecargas:
- Prefijo: `Contador& operator++()`
- Posfijo: `Contador operator++(int)`

El parámetro `int` de posfijo es un parámetro ficticio que permite distinguir la firma; normalmente no se utiliza para el cálculo.

### **5.3 Ejemplo completo de `++` y `--` de prefijo**

La clase `Contador` encapsula un entero privado. Los operadores `++` y `--` modifican ese atributo y devuelven una referencia al objeto actual.

`*this` representa el objeto actual. Devolver `Contador&` permite que la expresión represente el objeto ya modificado.

````python
%%writefile incremento_decremento.cpp
#include <iostream>
using namespace std;

class Contador {
private:
    int valor;

public:
    Contador(int inicial = 0) : valor(inicial) {}

    Contador& operator++() {
        ++valor;
        return *this;
    }

    Contador& operator--() {
        --valor;
        return *this;
    }

    int obtenerValor() const {
        return valor;
    }
};

int main() {
    Contador c(10);

    ++c;
    cout << "Despues de ++c: " << c.obtenerValor() << '\n';

    --c;
    cout << "Despues de --c: " << c.obtenerValor() << '\n';
    return 0;
}
````

### **5.4 Diferencia entre prefijo y posfijo**

Esta implementación permite ver que `c++` devuelve una copia del valor anterior, mientras que `++c` devuelve el objeto ya incrementado.

- `Contador anterior = c++;` guarda en `anterior` el valor previo al incremento.
- `Contador& actualizado = ++c;` hace que `actualizado` sea una referencia al objeto incrementado.

````python
%%writefile prefijo_posfijo.cpp
#include <iostream>
using namespace std;

class Contador {
private:
    int valor;

public:
    Contador(int inicial = 0) : valor(inicial) {}

    Contador& operator++() { // Prefijo
        ++valor;
        return *this;
    }

    Contador operator++(int) { // Posfijo
        Contador anterior = *this;
        valor++;
        return anterior;
    }

    int obtenerValor() const {
        return valor;
    }
};

int main() {
    Contador c(5);

    Contador anterior = c++;
    cout << "Valor devuelto por c++: " << anterior.obtenerValor() << '\n';
    cout << "Valor actual de c: " << c.obtenerValor() << '\n';

    Contador& actualizado = ++c;
    cout << "Valor devuelto por ++c: " << actualizado.obtenerValor() << '\n';
    cout << "Valor actual de c: " << c.obtenerValor() << '\n';
    return 0;
}
````

# **6. Sobrecarga de operadores de entrada y salida**

C++ utiliza flujos para intercambiar información:
- `cout` es un flujo de salida estándar.
- `cin` es un flujo de entrada estándar.
- `<<` envía información al flujo.
- `>>` extrae información del flujo.

Para permitir expresiones como `cout << objeto` y `cin >> objeto`, normalmente se definen funciones externas a la clase:
- `operator<<` recibe el flujo y el objeto a mostrar.
- `operator>>` recibe el flujo y una referencia al objeto que se va a modificar.

Se devuelve el flujo (`ostream&` o `istream&`) para permitir encadenar operaciones. El parámetro de salida suele ser una referencia constante al objeto, porque solo lo consulta; el de entrada es una referencia no constante, porque lo modifica.

### **6.1 Sobrecarga de `<<`: mostrar objetos**

La función `operator<<` recibe el flujo `ostream` como primer parámetro y el objeto `Punto` como segundo. Se implementa fuera de la clase porque el operando izquierdo de `cout << p` es el flujo.

Al devolver `ostream&`, se permite encadenar expresiones como `cout << p1 << p2`.

````python
%%writefile operador_salida.cpp
#include <iostream>
using namespace std;

class Punto {
public:
    int x, y;
    Punto(int x, int y) : x(x), y(y) {}
};

ostream& operator<<(ostream& salida, const Punto& p) {
    salida << "(" << p.x << ", " << p.y << ")";
    return salida;
}

int main() {
    Punto p1(3, 5);
    Punto p2(8, 2);

    cout << "Primer punto: " << p1 << '\n';
    cout << "Segundo punto: " << p2 << '\n';
    cout << "Ambos puntos: " << p1 << " y " << p2 << '\n';
    return 0;
}
````

### **6.2 Sobrecarga de `>>`: leer objetos**

La función `operator>>` recibe el flujo `istream` y una referencia no constante al objeto. Los datos que introduce el usuario se guardan en los atributos del objeto.

El ejemplo siguiente es interactivo: al ejecutarlo, introduce dos enteros separados por un espacio, por ejemplo `7 9`.

````python
%%writefile operador_entrada.cpp
#include <iostream>
using namespace std;

class Punto {
public:
    int x, y;
    Punto(int x = 0, int y = 0) : x(x), y(y) {}
};

istream& operator>>(istream& entrada, Punto& p) {
    entrada >> p.x >> p.y;
    return entrada;
}

ostream& operator<<(ostream& salida, const Punto& p) {
    salida << "(" << p.x << ", " << p.y << ")";
    return salida;
}

int main() {
    Punto p;
    cout << "Introduce las coordenadas x e y: ";
    cin >> p;
    cout << "Punto capturado: " << p << '\n';
    return 0;
}
````

# **7. Ejemplo integrador: clase `Fraccion`**

Este programa combina varias ideas:
- `operator+` suma dos fracciones.
- `operator==` compara dos fracciones matemáticamente, incluso si están escritas con numeradores y denominadores diferentes.
- `operator<<` muestra una fracción.

Para sumar `a/b + c/d`, se utiliza `(a*d + c*b)/(b*d)`. Para comparar fracciones, se comparan los productos cruzados. El ejemplo asume denominadores distintos de cero y no reduce automáticamente las fracciones.

### **7.1 Sumar y comparar fracciones**

Observa cómo una misma clase puede ofrecer operaciones naturales mediante operadores sobrecargados. La implementación de `==` compara los productos cruzados, de modo que `1/2` y `2/4` se consideran iguales aunque sus atributos sean distintos.

````python
%%writefile fraccion_integrador.cpp
#include <iostream>
using namespace std;

class Fraccion {
public:
    int numerador;
    int denominador;

    Fraccion(int n, int d) : numerador(n), denominador(d) {}

    Fraccion operator+(const Fraccion& otra) const {
        return Fraccion(
            numerador * otra.denominador + otra.numerador * denominador,
            denominador * otra.denominador
        );
    }

    bool operator==(const Fraccion& otra) const {
        return numerador * otra.denominador ==
               otra.numerador * denominador;
    }
};

ostream& operator<<(ostream& salida, const Fraccion& f) {
    salida << f.numerador << "/" << f.denominador;
    return salida;
}

int main() {
    Fraccion a(1, 2);
    Fraccion b(1, 3);
    Fraccion suma = a + b;
    Fraccion c(2, 4);

    cout << a << " + " << b << " = " << suma << '\n';
    cout << boolalpha;
    cout << "1/2 es igual a 2/4: " << (a == c) << '\n';
    return 0;
}
````
