# SESIÓN 1 — Fundamentos y operaciones con cadenas

##**1. ¿Qué es una cadena de caracteres?**

Una cadena es una secuencia de caracteres que representa texto, por ejemplo `"Hola"`, `"Programacion"` o `"Universidad"`.

En C, las cadenas se almacenan en arreglos de tipo `char` y terminan con el carácter especial `\0` (escrito en C como `\\0`).

**Usos comunes:** almacenar nombres, comparar palabras, construir mensajes y procesar texto.

##**2. Representación de una cadena en C**

```c
char palabra[] = "Hola";
```

Conceptualmente, el arreglo contiene:

| Índice | 0 | 1 | 2 | 3 | 4 |
|---|---|---|---|---|---|
| Contenido | H | o | l | a | `\0` |

- Los índices comienzan en cero.
- `\0` marca el final de la cadena.
- El terminador no forma parte de la longitud del texto.

##**3. Declaración y capacidad**

```c
char palabra[] = "Computadora"; // el compilador calcula el tamaño
char nombre[30];               // capacidad total de 30 caracteres
char ciudad[20] = "Juarez";    // texto inicial y espacio disponible
```

**Longitud** es la cantidad de caracteres del texto. **Capacidad** es el espacio reservado en el arreglo. Para guardar una cadena de `n` caracteres se necesitan, como mínimo, `n + 1` posiciones: una adicional para `\0`.

## **4. Leer cadenas: `scanf()` y `fgets()`**

Para una sola palabra podemos usar un ancho máximo en `scanf()`:

```c
char nombre[30];
scanf("%29s", nombre);
```

`scanf("%29s", nombre)` se detiene en el primer espacio. Para leer una línea que puede contener espacios, podemos usar:

```c
fgets(nombre, sizeof(nombre), stdin);
```

`fgets()` puede conservar el salto de línea `\n` si cabe en el arreglo.

## **5. Longitud con `strlen()`**

La función `strlen()` pertenece a `<string.h>` y cuenta los caracteres anteriores a `\0`.

```c
strlen(cadena);
```

| Cadena | Longitud |
|---|---:|
| `"Hola"` | 4 |
| `"Escuela"` | 7 |
| `"Computadora"` | 11 |
| `"Juarez"` | 6 |

### Ejemplo ejecutable: calcular longitud

````python
%%writefile longitud.c
#include <stdio.h>
#include <string.h>

int main(void) {
    char palabra[100];

    printf("Escribe una palabra: ");
    if (scanf("%99s", palabra) != 1) {
        return 1;
    }

    printf("Palabra: %s\n", palabra);
    printf("Longitud: %zu\n", strlen(palabra));
    return 0;
}
````

## **6. Diferencia entre `strlen()` y `sizeof()`**

- `strlen(cadena)`: cuenta los caracteres del texto, sin incluir `\0`.
- `sizeof(arreglo)`: devuelve el tamaño en bytes del arreglo completo.

```c
char palabra[20] = "Hola";
```

En este caso, `strlen(palabra)` devuelve `4` y `sizeof(palabra)` devuelve `20`.

**Precaución:** dentro de una función, un parámetro declarado como arreglo se trata como puntero; `sizeof` aplicado a ese parámetro no recupera la capacidad del arreglo original.

````python
%%writefile longitud_sizeof.c
#include <stdio.h>
#include <string.h>

int main(void) {
    char palabra[20] = "Hola";
    printf("Longitud del texto: %zu\n", strlen(palabra));
    printf("Tamano del arreglo: %zu bytes\n", sizeof(palabra));
    return 0;
}
````

## **7. Comparar cadenas con `strcmp()`**

`strcmp()` compara dos cadenas y devuelve:
- un valor menor que cero si la primera precede a la segunda;
- cero si ambas cadenas son iguales;
- un valor mayor que cero si la primera sigue a la segunda.

Para comprobar igualdad se utiliza:

```c
strcmp(cadena1, cadena2) == 0
```

La comparación distingue mayúsculas y minúsculas: `"Perro"` y `"perro"` son diferentes.

````python
%%writefile comparar.c
#include <stdio.h>
#include <string.h>

int main(void) {
    char cadena1[50];
    char cadena2[50];

    printf("Primera palabra: \n");
    if (scanf("%49s", cadena1) != 1) return 1;

    printf("Segunda palabra: \n");
    if (scanf("%49s", cadena2) != 1) return 1;

    if (strcmp(cadena1, cadena2) == 0) {
        printf("Las palabras son iguales.\n");
    } else {
        printf("Las palabras son diferentes.\n");
    }
    return 0;
}
````

````python
!gcc comparar.c -o comparar
!printf "Perro\nperro\n" | ./comparar
````

## **8. ¿Por qué no comparar cadenas con `==`?**

`==` no compara el contenido textual de dos arreglos. Para comparar el texto, usa `strcmp()`.

**Incorrecto para comparar contenido:**
```c
if (cadena1 == cadena2) { /* ... */ }
```

**Correcto:**
```c
if (strcmp(cadena1, cadena2) == 0) { /* ... */ }
```

## **9. Copiar y concatenar: `strcpy()` y `strcat()`**

**Copiar** reemplaza el contenido del destino con el contenido del origen:

```c
strcpy(destino, origen);
```

**Concatenar** agrega el origen al final del texto que ya existe en el destino:

```c
strcat(destino, origen);
```

Ejemplo: si `saludo` contiene `"Hola "` y `nombre` contiene `"Ana"`, después de `strcat(saludo, nombre)`, `saludo` contiene `"Hola Ana"`.

**Seguridad:** el arreglo de destino debe tener suficiente espacio para el texto resultante y `\0`. `strcpy()` y `strcat()` no comprueban automáticamente la capacidad.

##**10. Ejemplo ejecutable: construir un nombre completo**

````python
%%writefile nombre_completo.c
#include <stdio.h>
#include <string.h>

int main(void) {
    char nombre[50];
    char apellido[50];
    char nombreCompleto[110];

    printf("Nombre: \n");
    if (scanf("%49s", nombre) != 1) return 1;

    printf("Apellido: \n");
    if (scanf("%49s", apellido) != 1) return 1;

    strcpy(nombreCompleto, nombre);
    strcat(nombreCompleto, " ");
    strcat(nombreCompleto, apellido);

    printf("\n");
    printf("Nombre completo: %s\n", nombreCompleto);
    printf("Longitud del nombre completo: %zu\n",
           strlen(nombreCompleto));
    return 0;
}
````

````python
!gcc nombre_completo.c -o nombre_completo
!printf "Ana\nMartinez\n" | ./nombre_completo
````

## **11. Buscar caracteres con `strchr()`**

`strchr()` busca la primera aparición de un carácter dentro de una cadena.

```c
char *resultado = strchr(cadena, caracter);
```

La función devuelve un puntero a la primera aparición si la encuentra; si no, devuelve `NULL`.

Ejemplo:

```c
char palabra[] = "Computadora";
char *resultado = strchr(palabra, 'o');
```

La búsqueda distingue mayúsculas y minúsculas.

````python
%%writefile buscar_caracter.c
#include <stdio.h>
#include <string.h>

int main(void) {
    char palabra[100];
    char caracter;
    char *resultado;

    printf("Palabra: \n");
    if (scanf("%99s", palabra) != 1) return 1;

    printf("Caracter a buscar: \n");
    if (scanf(" %c", &caracter) != 1) return 1;

    printf("\n");
    resultado = strchr(palabra, caracter);

    if (resultado != NULL) {
        printf("Caracter encontrado.\n");
        printf("Primera posicion (indice desde cero): %td\n",
               resultado - palabra);
    } else {
        printf("Caracter no encontrado.\n");
    }

    return 0;
}
````

````python
!gcc buscar_caracter.c -o buscar_caracter
!printf "Computadora\no\n" | ./buscar_caracter
````

## **12. Interpretar `strchr()` y calcular la posición**

- Si `resultado != NULL`, el carácter fue encontrado.
- Si `resultado == NULL`, no fue encontrado.
- Si se encontró, `resultado - palabra` calcula el índice de la primera aparición.
- Los índices empiezan en cero.

En `"Computadora"`, la primera `t` está en el índice `5`.

**Importante:** comprueba que el resultado no sea `NULL` antes de restar punteros o utilizarlo.

## **13. Recorrer una cadena con `for`**

Para procesar todos los caracteres, se puede utilizar este patrón:

```c
for (int i = 0; palabra[i] != '\0'; i++) {
    printf("%c\n", palabra[i]);
}
```

El ciclo:
1. comienza en el índice cero;
2. continúa mientras no encuentre `\0`;
3. procesa el carácter actual;
4. avanza a la siguiente posición.

Este patrón permite contar vocales, contar caracteres o analizar cada posición.

````python
%%writefile recorrer.c
#include <stdio.h>

int main(void) {
    char palabra[] = "Hola";

    for (int i = 0; palabra[i] != '\0'; i++) {
        printf("Indice %d: %c\n", i, palabra[i]);
    }

    return 0;
}
````

## **14. Algoritmo para contar vocales**

1. Leer la palabra.
2. Inicializar `vocales = 0`.
3. Recorrer cada carácter.
4. Convertirlo a minúscula para reconocer también las mayúsculas.
5. Si es `a`, `e`, `i`, `o` o `u`, incrementar el contador.
6. Mostrar el total.

**Pseudocódigo:**
```text
LEER palabra
vocales = 0
PARA cada caracter hasta ' '
    SI el caracter es una vocal
        vocales = vocales + 1
MOSTRAR vocales
```

## **15. Identificar vocales con `if`, `||` y `tolower()`**

El operador `||` significa **OR** (o lógico).

```c
if (caracter == 'a' || caracter == 'e' ||
    caracter == 'i' || caracter == 'o' ||
    caracter == 'u') {
    vocales++;
}
```

Para reconocer vocales mayúsculas, se puede normalizar la letra con `tolower()` de `<ctype.h>`:

```c
char letra = (char)tolower((unsigned char)caracter);
```

Después, se compara `letra` con las vocales minúsculas.

**Nota:** este ejemplo reconoce las cinco vocales básicas sin acento. Las vocales acentuadas requieren tratar adecuadamente la codificación del texto.

````python
%%writefile contar_vocales.c
#include <stdio.h>
#include <ctype.h>

int main(void) {
    char palabra[100];
    int vocales = 0;

    printf("Palabra: \n");
    if (scanf("%99s", palabra) != 1) return 1;

    printf("\n");
    for (int i = 0; palabra[i] != '\0'; i++) {
        char caracter = (char)tolower((unsigned char)palabra[i]);

        if (caracter == 'a' || caracter == 'e' ||
            caracter == 'i' || caracter == 'o' ||
            caracter == 'u') {
            vocales++;
        }
    }

    printf("Cantidad de vocales: %d\n", vocales);
    return 0;
}
````
