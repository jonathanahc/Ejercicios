# **Apuntadores en C**

## **1. Variables, contenido y dirección de memoria**


Una variable tiene un **nombre**, un **tipo**, un **valor** y una **dirección de memoria**. El valor y la dirección no son lo mismo.

- `numero` permite acceder al valor almacenado en la variable.
- `&numero` obtiene la dirección de esa variable.
- Las direcciones concretas pueden cambiar entre ejecuciones; no debemos memorizar una dirección específica.

### **Ejemplo 1: consultar el valor y la dirección**

````python
%%writefile ejemplo1.c
#include <stdio.h>

int main(void) {
    int numero = 25;

    printf("Valor de numero: %d\n", numero);
    printf("Direccion de numero: %p\n", (void *)&numero);

    return 0;
}
````

## **2. ¿Qué es un apuntador?**


Un **apuntador** es una variable que almacena la dirección de un objeto en memoria. Así podemos acceder al objeto indirectamente.

Observa estas instrucciones:

```c
int numero = 25;
int *ptr = &numero;
```

- `numero` guarda el entero `25`.
- `&numero` obtiene la dirección de `numero`.
- `ptr` guarda esa dirección.
- `int *ptr` declara un apuntador a un objeto de tipo `int`.

Un mapa conceptual sencillo:

```text
numero: contiene 25
   ▲
   │ ptr almacena la dirección de numero
   │
ptr: contiene la dirección de numero
```

**Importante:** el apuntador almacena una dirección, no el valor `25` directamente.

### **Ejemplo 2: observar la relación entre una variable y un apuntador**

````python
%%writefile ejemplo2.c
#include <stdio.h>

int main(void) {
    int numero = 1500;
    int *ptr = &numero;

    printf("Valor de numero: %d\n", numero);
    printf("Direccion de numero: %p\n", (void *)&numero);
    printf("Direccion almacenada en ptr: %p\n", (void *)ptr);

    return 0;
}
````

## **3. Los operadores `&` y `*`**

En esta clase usaremos dos operadores con funciones diferentes:

- **`&` (dirección):** obtiene la dirección de un objeto, por ejemplo `&numero`.
- **`*` (desreferencia):** permite acceder al objeto señalado por un apuntador, por ejemplo `*ptr`.

El símbolo `*` también aparece en la declaración `int *ptr`; ahí indica que `ptr` es un apuntador. En una expresión como `*ptr`, se utiliza para acceder al objeto apuntado.

### **Ejemplo 3: leer el valor a través del apuntador**

````python
%%writefile ejemplo3.c
#include <stdio.h>

int main(void) {
    int numero = 1546;
    int *ptr = &numero;

    printf("numero = %d\n", numero);
    printf("*ptr = %d\n", *ptr);

    return 0;
}
````

### **Ejemplo 4: modificar una variable mediante desreferencia**

La expresión `*ptr = 90;` modifica el objeto al que apunta `ptr`. Si `ptr` apunta a `numero`, entonces cambia la propia variable `numero`.

````python
%%writefile ejemplo4.c
#include <stdio.h>

int main(void) {
    int numero = 50;
    int *ptr = &numero;

    printf("Antes: numero = %d\n", numero);

    *ptr = 254;

    printf("Despues: numero = %d\n", numero);
    printf("Direccion de numero: %p\n", (void *)&numero);
    printf("Direccion almacenada en ptr: %p\n", (void *)ptr);

    return 0;
}
````

## **4. ¿Qué significan `x`, `&x`, `p`, `*p` y `&p`?**



Considera:

```c
int x = 15;
int *p = &x;
```

| Expresión | Qué representa |
|---|---|
| `x` | El valor almacenado en `x`. |
| `&x` | La dirección de `x`. |
| `p` | La dirección almacenada en el apuntador `p`. |
| `*p` | El valor del objeto al que apunta `p`. |
| `&p` | La dirección de la propia variable apuntadora `p`. |

La dirección de `p` no es lo mismo que la dirección que `p` guarda. Un apuntador también es una variable y, por tanto, tiene su propia dirección.

### **Ejemplo 5: imprimir las cinco expresiones**

````python
%%writefile ejemplo5.c
#include <stdio.h>

int main(void) {
    int x = 15;
    int *p = &x;

    printf("x   = %d\n", x);
    printf("&x  = %p\n", (void *)&x);
    printf("p   = %p\n", (void *)p);
    printf("*p  = %d\n", *p);
    printf("&p  = %p\n", (void *)&p);

    return 0;
}
````

## **5. Reasignar un apuntador**



Un apuntador puede cambiar la dirección que almacena para señalar a otro objeto compatible. Reasignar el apuntador no modifica automáticamente las variables.

En el ejemplo, `p` primero apunta a `a` y después apunta a `b`.

````python
%%writefile ejemplo6.c
#include <stdio.h>

int main(void) {
    int a = 10;
    int b = 20;
    int *p = &a;

    printf("Al principio, *p = %d\n", *p);

    p = &b;
    printf("Despues de cambiar p, *p = %d\n", *p);

    *p = 60;
    printf("a = %d\n", a);
    printf("b = %d\n", b);

    return 0;
}
````

## **6. Aritmética de apuntadores y arreglos**



En C, el nombre de un arreglo suele convertirse en un apuntador a su primer elemento cuando se usa en una expresión. Por eso podemos recorrer un arreglo mediante aritmética de apuntadores.

Si `p` apunta al primer elemento:
- `*p` accede al primer elemento.
- `*(p + 1)` accede al segundo.
- `*(p + 2)` accede al tercero.
- `p++` avanza al siguiente elemento.

`p + 1` avanza un elemento del tipo apuntado, no necesariamente un byte.

**Seguridad:** la aritmética de apuntadores debe mantenerse dentro del mismo arreglo o, como límite, en la posición inmediatamente posterior al último elemento. Esa posición final no debe desreferenciarse.

### **Ejemplo 7: recorrer un arreglo sin usar `[]` para acceder a los elementos**

````python
%%writefile ejemplo7.c
#include <stdio.h>

int main(void) {
    int numeros[] = {10, 20, 30, 40, 50};
    int *p = numeros;
    int suma = 0;

    for (int i = 0; i < 5; i++) {
        printf("Valor: %d | Direccion: %p\n", *(p + i), (void *)(p + i));
        suma += *(p + i);
    }

    printf("Suma = %d\n", suma);

    return 0;
}
````

## **7. Mapas de memoria: seguir los cambios paso a paso**



Un mapa de memoria es un esquema que muestra las variables, sus contenidos y las relaciones entre los apuntadores y los objetos a los que señalan. En los ejercicios usaremos direcciones simbólicas —A, B, C— porque las direcciones reales las determina el entorno de ejecución.

Analiza este código:

```c
int a = 10;
int b = 20;
int c = 30;
int *p = &a;

*p = 50;
p = &b;
*p = 80;
p = &c;
*p = 100;
```

El estado final es:

| Variable | Valor final |
|---|---:|
| `a` | 50 |
| `b` | 80 |
| `c` | 100 |
| `p` | Guarda la dirección de `c` |

La instrucción `*p = ...` modifica el objeto apuntado. La instrucción `p = &...` cambia a qué objeto apunta `p`.

## **8. Errores frecuentes y buenas prácticas**



- **Desreferenciar un apuntador no inicializado:** puede provocar comportamiento indefinido. Inicialízalo antes de usar `*p`.
- **Confundir `p` con `*p`:** `p` es la dirección almacenada; `*p` es el objeto al que se accede.
- **Confundir `&p` con `p`:** `&p` es la dirección de la propia variable apuntadora.
- **Apuntar a un tipo incompatible:** utiliza apuntadores del tipo adecuado.
- **Salir de los límites de un arreglo:** no accedas a posiciones fuera del arreglo ni desreferencies la posición posterior al último elemento.
- **Imprimir direcciones con un formato incorrecto:** usa `%p` y convierte la expresión a `(void *)`.
