# **Clase 11 — Herencia y modificadores de acceso en C++**

##**Activación de conocimientos previos**

Antes de estudiar herencia, recuerda estos conceptos:

- **Clase:** definición que reúne datos y operaciones.
- **Objeto:** instancia concreta de una clase.
- **Atributo:** dato que describe el estado del objeto.
- **Método:** operación que puede realizar el objeto.
- **Encapsulación:** organización y protección del estado interno.
- **Constructor:** función especial que se ejecuta al crear un objeto.

Ejemplo: una clase `Persona` puede tener los atributos `nombre` y `edad`, además de un método `presentarse()`.

**¿Qué hacemos cuando varias clases necesitan los mismos datos y operaciones, pero también tienen características propias?**

## **1. El problema que da origen a la herencia**

Imagina que una universidad quiere registrar estudiantes, profesores y personal de laboratorio. Todos tienen nombre, edad e identificador, pero cada grupo posee datos específicos.

| Tipo de persona | Datos compartidos | Datos específicos |
|---|---|---|
| Estudiante | Nombre, edad, identificador | Matrícula, carrera |
| Profesor | Nombre, edad, identificador | Número de empleado, especialidad |
| Técnico de laboratorio | Nombre, edad, identificador | Laboratorio asignado, certificación |


Si cada clase define por separado los datos compartidos, repetimos código. Esa repetición puede volver más difícil el mantenimiento: si la forma de representar el identificador cambia, tendríamos que modificar varias clases.

### **Una solución inicial: clases independientes**

Observa que `nombre` y `edad` se repiten.

````python
#include <iostream>
#include <string>
using namespace std;

class EstudianteIndependiente {
public:
    string nombre;
    int edad;
    string matricula;
};

class ProfesorIndependiente {
public:
    string nombre;
    int edad;
    int numeroEmpleado;
};

int main() {
    EstudianteIndependiente estudiante;
    estudiante.nombre = "Mara";
    estudiante.edad = 21;
    estudiante.matricula = "IA-204";

    ProfesorIndependiente profesor;
    profesor.nombre = "Diego";
    profesor.edad = 39;
    profesor.numeroEmpleado = 731;

    cout << estudiante.nombre << " / " << estudiante.matricula << endl;
    cout << profesor.nombre << " / " << profesor.numeroEmpleado << endl;
    return 0;
}
````

## **2. La idea central: una relación entre clases**

La **herencia** es un mecanismo de POO que permite definir una clase a partir de otra, reutilizando sus miembros accesibles y agregando o especializando características.

- **Clase base:** la clase general de la que se deriva otra.
- **Clase derivada:** la clase especializada que parte de la clase base.
- **Reutilización:** aprovechar atributos y métodos existentes sin volver a declararlos.
- **Especialización:** incorporar características o comportamientos propios.

En nuestro sistema:
- `Persona` será la clase base.
- `Estudiante`, `Profesor` y `TecnicoLaboratorio` pueden ser clases derivadas.

La relación conceptual es “un estudiante es una persona”. Esta relación *es-un* suele ser una señal de que la herencia puede tener sentido.

### Representación del diseño

```text
                         Persona
                   _________|_________
                  |         |         |
             Estudiante  Profesor  TecnicoLaboratorio
                  |         |         |
              matricula  empleado   laboratorio
              carrera    especialidad certificacion
```

La clase base guarda lo común; cada derivada agrega lo específico. No se pretende que las clases derivadas sean idénticas, sino que compartan una estructura general.

## **3. Declarar una clase base**


Empecemos con `Persona`. Por ahora, los atributos son públicos para concentrarnos en la estructura de la herencia. Más adelante mejoraremos el diseño con `protected` y `private`.

````python
%%writefile programa1.cpp
#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;
    int edad;

    void presentarse() {
        cout << "Nombre: " << nombre << endl;
        cout << "Edad: " << edad << endl;
    }
};

int main() {
    Persona persona;
    persona.nombre = "Mara";
    persona.edad = 21;
    persona.presentarse();
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

La clase `Persona` define dos atributos y un método. Todavía no hay herencia; solo hemos creado la clase general que después servirá de base.

**Siguiente paso:** crear una clase `Estudiante` que utilice esa estructura común y agregue una matrícula.

## **4. Declarar la clase derivada**

En C++, la sintaxis general es:

```cpp
class ClaseDerivada : public ClaseBase {
    // Miembros propios
};
```

Ahora `Estudiante` deriva públicamente de `Persona`.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;
    int edad;

    void presentarse() {
        cout << "Nombre: " << nombre << endl;
        cout << "Edad: " << edad << endl;
    }
};

class Estudiante : public Persona {
public:
    string matricula;

    void mostrarMatricula() {
        cout << "Matricula: " << matricula << endl;
    }
};

int main() {
    Estudiante alumno;
    alumno.nombre = "Mara";       // Miembro público heredado
    alumno.edad = 21;             // Miembro público heredado
    alumno.matricula = "IA-204";  // Miembro propio

    alumno.presentarse();         // Método heredado
    alumno.mostrarMatricula();    // Método propio
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

## **5. Ampliar el diseño: varias clases derivadas**

La ventaja se aprecia mejor cuando varias clases reutilizan la misma clase base. Vamos a agregar un profesor.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;
    int edad;

    Persona(string n, int e) : nombre(n), edad(e) {}

    void presentarse() {
        cout << nombre << ", " << edad << " anios" << endl;
    }
};

class Estudiante : public Persona {
public:
    string matricula;

    Estudiante(string n, int e, string m)
        : Persona(n, e), matricula(m) {}
};

class Profesor : public Persona {
public:
    int numeroEmpleado;
    string especialidad;

    Profesor(string n, int e, int num, string esp)
        : Persona(n, e), numeroEmpleado(num), especialidad(esp) {}
};

int main() {
    Estudiante estudiante("Mara", 21, "IA-204");
    Profesor profesor("Diego", 39, 731, "Robotica");

    estudiante.presentarse();
    cout << "Matricula: " << estudiante.matricula << endl;

    profesor.presentarse();
    cout << "Empleado: " << profesor.numeroEmpleado << endl;
    cout << "Especialidad: " << profesor.especialidad << endl;
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

## **6. Un nuevo problema: ¿todos deben modificar todos los datos?**


Hasta aquí declaramos los atributos como `public`. Eso hace que el código externo pueda modificarlos directamente:

```cpp
alumno.edad = -500;
```

El programa lo permitiría aunque el valor no tenga sentido. Para controlar cómo se utiliza la información, C++ ofrece modificadores de acceso.

Los tres modificadores principales son:
- `public`
- `protected`
- `private`

## **7. `public`: interfaz accesible desde el exterior**



Los miembros públicos pueden utilizarse desde fuera de la clase a través de un objeto, si el código tiene acceso a ese objeto. También pueden utilizarse desde las clases derivadas.

Ejemplo:
```cpp
class Persona {
public:
    string nombre;
};
```

Uso:
```cpp
Persona p;
p.nombre = "Mara";
```

## **8. `protected`: compartir miembros con las clases derivadas**

Supongamos que queremos que una clase derivada pueda utilizar la edad, pero no permitir el acceso directo desde `main()`. Podemos declarar `edad` como `protected`.

Ejecuta este ejemplo:

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;

protected:
    int edad;

public:
    Persona(string n, int e) : nombre(n), edad(e) {}
};

class Estudiante : public Persona {
public:
    string matricula;

    Estudiante(string n, int e, string m)
        : Persona(n, e), matricula(m) {}

    void mostrarEdad() {
        cout << "Edad: " << edad << endl; // Permitido en la derivada
    }

    void cumplirAnios() {
        edad++;
    }
};

int main() {
    Estudiante alumno("Mara", 21, "IA-204");
    alumno.mostrarEdad();
    alumno.cumplirAnios();
    alumno.mostrarEdad();

    // cout << alumno.edad; // ERROR: protected no permite acceso directo desde main()
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````


**Precaución:** `protected` puede acoplar las clases derivadas a los detalles internos de la base. En sistemas grandes, a veces es preferible mantener los atributos privados y proporcionar métodos controlados.

## **9. `private`: proteger datos internos**

Ahora imaginemos que cada persona tiene un identificador confidencial. No queremos que las clases derivadas lean o modifiquen ese dato directamente. Lo declaramos `private` y ofrecemos un método para validar una entrada.

La clase derivada puede utilizar el método público, pero no acceder directamente al atributo privado.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
private:
    string identificador;

public:
    string nombre;

    Persona(string n, string id)
        : identificador(id), nombre(n) {}

    bool validarIdentificador(string intento) {
        return intento == identificador;
    }
};

class Estudiante : public Persona {
public:
    string matricula;

    Estudiante(string n, string id, string m)
        : Persona(n, id), matricula(m) {}

    void comprobarIdentificador(string intento) {
        if (validarIdentificador(intento)) {
            cout << "Identificador validado" << endl;
        } else {
            cout << "Identificador incorrecto" << endl;
        }

        // cout << identificador; // ERROR: es private en Persona
    }
};

int main() {
    Estudiante alumno("Mara", "ID-884", "IA-204");
    alumno.comprobarIdentificador("ID-884");
    alumno.comprobarIdentificador("ID-000");
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

## **10. Miembros heredados: qué puede utilizar una derivada**

Consideremos una clase base con los tres niveles de acceso:

```cpp
class Dispositivo {
public:
    string modelo;
protected:
    int nivelBateria;
private:
    string claveInterna;
};
```

Si `RobotExplorador` hereda públicamente de `Dispositivo`:
- Puede utilizar directamente `modelo`.
- Puede utilizar directamente `nivelBateria`.
- No puede utilizar directamente `claveInterna`.
- Puede utilizar métodos públicos o protegidos de `Dispositivo` que operen con `claveInterna`, si existen.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Dispositivo {
public:
    string modelo;

protected:
    int nivelBateria;

private:
    string claveInterna;

public:
    Dispositivo(string m, int bateria, string clave)
        : modelo(m), nivelBateria(bateria), claveInterna(clave) {}

    bool autenticar(string clave) {
        return clave == claveInterna;
    }
};

class RobotExplorador : public Dispositivo {
public:
    RobotExplorador(string m, int bateria, string clave)
        : Dispositivo(m, bateria, clave) {}

    void consumirEnergia(int cantidad) {
        nivelBateria -= cantidad; // Permitido: protected
    }

    void mostrarEstado() {
        cout << "Modelo: " << modelo << endl;
        cout << "Bateria: " << nivelBateria << "%" << endl;
    }
};

int main() {
    RobotExplorador robot("Nereida-X", 90, "R-77");
    robot.consumirEnergia(15);
    robot.mostrarEstado();
    cout << "Autenticacion: " << robot.autenticar("R-77") << endl;
    // cout << robot.claveInterna; // ERROR: private en Dispositivo
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

Para que este objeto exista correctamente, se debe construir tanto la parte `Dispositivo` como la parte `RobotExplorador`. Eso nos lleva a los constructores en herencia.

## **11. Constructores en herencia: el orden importa**

Al crear un objeto derivado, C++ inicializa primero la parte de la clase base y después los miembros propios de la clase derivada. Una vez completadas esas inicializaciones, se ejecuta el cuerpo del constructor derivado.

```text
Crear RobotExplorador
        |
        v
Inicializar Dispositivo
        |
        v
Inicializar miembros propios de RobotExplorador
        |
        v
Ejecutar cuerpo del constructor derivado
        |
        v
Objeto listo
```

Esto garantiza que la parte base del objeto esté construida antes de que el constructor derivado dependa de ella.

## **12. Lista de inicialización del constructor**

La lista de inicialización aparece después de los dos puntos (`:`) en la definición del constructor. Permite inicializar miembros y elegir el constructor de la clase base.

En este ejemplo:

```cpp
Estudiante(string n, int e, string m)
    : Persona(n, e), matricula(m) {
}
```

- `Persona(n, e)` llama al constructor de la clase base.
- `matricula(m)` inicializa el atributo de la clase derivada.
- El cuerpo `{ }` puede quedar vacío si no se necesita realizar ninguna operación adicional.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;

    Persona(string n) : nombre(n) {
        cout << "1. Constructor de Persona" << endl;
    }
};

class Estudiante : public Persona {
public:
    string matricula;

    Estudiante(string n, string m)
        : Persona(n), matricula(m) {
        cout << "2. Constructor de Estudiante" << endl;
    }
};

int main() {
    cout << "Creando el objeto..." << endl;
    Estudiante alumno("Mara", "IA-204");
    cout << "Objeto creado: " << alumno.nombre << endl;
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

## **13. ¿Qué sucede si no se llama al constructor correcto de la base?**


Si la clase base solo tiene un constructor que exige argumentos y no existe un constructor sin parámetros accesible, el constructor derivado debe llamar explícitamente a uno apropiado.

El ejemplo incorrecto está comentado para que el programa siga compilando.

````python
%%writefile programa1.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;
    Persona(string n) : nombre(n) {}
};

class Estudiante : public Persona {
public:
    string matricula;

    // INCORRECTO:
    // Estudiante(string m) : matricula(m) {}
    // C++ intentaría construir Persona sin argumentos, pero no existe ese constructor.

    // CORRECTO:
    Estudiante(string n, string m)
        : Persona(n), matricula(m) {}
};

int main() {
    Estudiante alumno("Mara", "IA-204");
    cout << alumno.nombre << " / " << alumno.matricula << endl;
    return 0;
}
````

````python
!g++ programa1.cpp && ./a.out
````

**Regla práctica:** cuando la clase base requiere parámetros, pásalos desde la lista de inicialización del constructor derivado. No intentes inicializar los miembros de la base como si fueran atributos nuevos de la derivada.

## **14. Ejemplo integrador completo: expediente universitario**

Ahora reunimos todos los conceptos de la clase en un único programa:
- `Persona` es la clase base.
- `Estudiante` es la clase derivada.
- `nombre` es público.
- `edad` es protegido.
- `identificador` es privado.
- El constructor de `Estudiante` inicializa la base mediante `Persona(...)`.
- La derivada puede utilizar `edad`, pero no leer directamente `identificador`.
- La consulta del identificador se realiza mediante un método público de la base.

````python
%%writefile programa2.cpp

#include <iostream>
#include <string>
using namespace std;

class Persona {
public:
    string nombre;

protected:
    int edad;

private:
    string identificador;

public:
    Persona(string n, int e, string id)
        : nombre(n), edad(e), identificador(id) {
        cout << "Construyendo Persona" << endl;
    }

    string obtenerIdentificador() {
        return identificador;
    }
};

class Estudiante : public Persona {
public:
    string matricula;
    string carrera;

    Estudiante(string n, int e, string id, string m, string c)
        : Persona(n, e, id), matricula(m), carrera(c) {
        cout << "Construyendo Estudiante" << endl;
    }

    void mostrarExpediente() {
        cout << "\n--- Expediente ---" << endl;
        cout << "Nombre: " << nombre << endl;
        cout << "Edad: " << edad << endl;
        cout << "Identificador: " << obtenerIdentificador() << endl;
        cout << "Matricula: " << matricula << endl;
        cout << "Carrera: " << carrera << endl;
    }

    void cumplirAnios() {
        edad++;
    }
};

int main() {
    Estudiante alumno("Mara Solis", 21, "ID-884", "IA-204", "Inteligencia Artificial");
    alumno.mostrarExpediente();

    cout << "\nActualizando edad..." << endl;
    alumno.cumplirAnios();
    alumno.mostrarExpediente();

    // alumno.edad = 22;             // ERROR: protected
    // alumno.identificador = "X";   // ERROR: private
    return 0;
}
````

````python
!g++ programa2.cpp && ./a.out
````

### Análisis del programa

1. `Persona` reúne información común.
2. `Estudiante` reutiliza la parte pública y protegida de la base.
3. `identificador` permanece privado y se consulta mediante `obtenerIdentificador()`.
4. La lista `: Persona(n, e, id)` ejecuta el constructor de la base con los argumentos requeridos.
5. La salida permite observar que el constructor de `Persona` se ejecuta antes que el de `Estudiante`.

Esta es la idea central de la clase: la herencia reutiliza características, pero los modificadores de acceso continúan controlando cómo se usan.
